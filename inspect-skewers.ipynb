```python
import numpy as np
import matplotlib.pyplot as plt
import h5py
```

```python
fname = 'skewers.h5'
with h5py.File(fname, 'r') as f:

  for key, value in f.attrs.items():
    print(f'{key} : {value}')

  s_HI = np.asarray(f['HI_density'])
  print(s_HI.shape)
  s_T  = np.asarray(f['temperature'])
  s_v  = np.asarray(f['velocity'])
```

```python
plt.plot(s_HI[0,:])
plt.plot(s_HI[10,:])

plt.yscale('log')
```

```python
plt.plot(s_T[0,:])
plt.plot(s_T[10,:])

plt.yscale('log')
```

```python
plt.plot(s_v[0,:])
plt.plot(s_v[10,:])
```
